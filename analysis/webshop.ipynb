# WebShop analysis

Uses `stateact.analysis` and the unified result format.

```python
from pathlib import Path
from stateact.analysis import summarise

repo = Path.cwd()
if (repo / "webshop_runs").is_dir():
    root = repo / "webshop_runs"
elif (repo.parent / "webshop_runs").is_dir():
    root = repo.parent / "webshop_runs"
else:
    root = repo

run = root / "results" / "paper_repro_14b"
runs = summarise(run)
runs
```

```python
# Best agent per model
runs.sort_values("score").groupby("model").tail(1).sort_values(["model", "score"])
```
