# ALFWorld analysis

Uses `stateact.analysis` and the unified `episodes.csv` / `summary.csv` format.

```python
from pathlib import Path
from stateact.analysis import load_episodes, summarise

repo = Path.cwd()
if (repo / "alfworld_runs").is_dir():
    root = repo / "alfworld_runs"
elif (repo.parent / "alfworld_runs").is_dir():
    root = repo.parent / "alfworld_runs"
else:
    root = repo

run = root / "results" / "paper_repro_14b_fixed"
episodes = load_episodes(run)
print(f"{len(episodes)} episodes")
episodes[["env_idx", "env_type", "success", "done", "steps", "early_stop"]].head()
```

```python
episodes.groupby("env_type")["success"].agg(["count", "sum", "mean"])
```
