# TextCraft analysis

Uses the shared `stateact.analysis` helpers and the unified result format.

```python
from pathlib import Path
from stateact.analysis import textcraft

repo = Path.cwd()
if (repo / "textcraft_runs").is_dir():
    textcraft_dir = repo / "textcraft_runs"
elif (repo.parent / "textcraft_runs").is_dir():
    textcraft_dir = repo.parent / "textcraft_runs"
else:
    textcraft_dir = repo

run = textcraft_dir / "results" / "stateact_repro"
frame = textcraft.load_results(run)
textcraft.summarise_results(frame)
```
